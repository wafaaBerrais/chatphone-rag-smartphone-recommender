vectorisation et creation des index de recherche

```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
import faiss
from transformers import AutoTokenizer, AutoModel

def build_full_text(row):
    return (
        f"Brand: {row['brand']}. "
        f"Price: {row['prices']}. "
        f"Product: {row['title_x']}. "
        f"Review Title: {row['title_y']}. "
        f"Review: {row['body']}."
    )

df['full_text'] = df.apply(build_full_text, axis=1)

# modèle Sentence-BERT (pour la vectorisation)
model = SentenceTransformer('all-MiniLM-L6-v2')

# vectorisation en batch
texts = df['full_text'].tolist()
embeddings = model.encode(
    texts,
    batch_size=512,
    show_progress_bar=True,
    convert_to_numpy=True
).astype('float32')

# construction de l’index FAISS (pour la recherche sémentique)
dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(embeddings)
print(f"Index FAISS : {index.ntotal} vecteurs")

# fonction de recherche sémantique
def semantic_search(query: str, top_k: int = 5):
    q_vec = model.encode([query], convert_to_numpy=True).astype('float32')
    distances, indices = index.search(q_vec, k=top_k)
    results = df.iloc[indices[0]][
        ['asin','brand','title_x','prices','title_y','body']
    ].copy()
    results['distance'] = distances[0]
    return results

# exemple d’utilisation
if __name__ == "__main__":
    requete = "smartphone bonne caméra belles photos"
    top5 = semantic_search(requete, top_k=5)
    print(top5.to_string(index=False))
```

analyse sentimentale

```python
from sentence_transformers import SentenceTransformer
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
import numpy as np
import faiss
import pandas as pd

df         = pd.read_pickle(f'{CACHE_DIR}/df_fulltext.pkl')
embeddings = np.load(f'{CACHE_DIR}/embeddings.npy')            # shape = (N, D), dtype=float32
index_all  = faiss.read_index(f'{CACHE_DIR}/faiss_index.idx')
model      = SentenceTransformer('all-MiniLM-L6-v2')

# pré-calcul du score VADER sur chaque review
analyzer = SentimentIntensityAnalyzer()
df['sentiment_score'] = df['body'].fillna('').apply(
    lambda txt: analyzer.polarity_scores(txt)['compound']
)

# construction des deux sous‐indices FAISS (positif / négatif)
pos_idx = np.where(df['sentiment_score'] >  0.05)[0]
neg_idx = np.where(df['sentiment_score'] < -0.05)[0]

d         = embeddings.shape[1]
index_pos = faiss.IndexFlatL2(d)
index_neg = faiss.IndexFlatL2(d)
index_pos.add(embeddings[pos_idx])
index_neg.add(embeddings[neg_idx])

# fonction de recherche sémantique filtrée par sentiment
def semantic_search(query: str,
                    top_k: int = 5,
                    sentiment: str = None):
    """
    sentiment=None      → recherche sur l'index global
    sentiment='positive'→ uniquement sur reviews positives
    sentiment='negative'→ uniquement sur reviews négatives
    """

    q_emb = model.encode([query], convert_to_numpy=True).astype('float32')

    # choix de l'index
    if sentiment == 'positive':
        dists, idxs = index_pos.search(q_emb, top_k)
        sel = pos_idx[idxs[0]]
    elif sentiment == 'negative':
        dists, idxs = index_neg.search(q_emb, top_k)
        sel = neg_idx[idxs[0]]
    else:
        dists, idxs = index_all.search(q_emb, top_k)
        sel = idxs[0]

    # construction du DataFrame résultat
    res = df.iloc[sel].copy()
    res['distance']        = dists[0]
    return res[[
        'asin','brand','title_x','prices',
        'title_y','body','sentiment_score','distance'
    ]]

# test
if __name__ == "__main__":
    print("Top-5 reviews POSITIVES pour « bonne caméra » :")
    print(semantic_search("bonne caméra", top_k=5, sentiment='positive').to_string(index=False))
```

API pour la communication avec mon local

```python
from fastapi import FastAPI
from pydantic import BaseModel
from pyngrok import ngrok, conf
import pandas as pd
import faiss
from sentence_transformers import SentenceTransformer
import uvicorn
import nest_asyncio

nest_asyncio.apply()

# token ngrok
import os
conf.get_default().auth_token = os.environ["NGROK_AUTHTOKEN"]  # never hard-code the token

df        = pd.read_pickle(f'{CACHE_DIR}/df_fulltext.pkl')
index_all = faiss.read_index(f'{CACHE_DIR}/faiss_index_all.idx')
embedder  = SentenceTransformer('all-MiniLM-L6-v2')

# ouvre le tunnel ngrok
public_url = ngrok.connect(8000, bind_tls=True).public_url
print("API accessible via :", public_url)

# définis l’API
app = FastAPI()
class Query(BaseModel):
    text: str
    k:    int = 50

@app.post("/search")
def search(q: Query):
    # encode + recherche FAISS
    q_emb = embedder.encode([q.text])
    faiss.normalize_L2(q_emb)
    D, I  = index_all.search(q_emb, q.k)

    # construisez une liste d’objets
    rows    = df.iloc[I[0]]
    results = []
    for _, row in rows.iterrows():
        results.append({
            "brand":  row["brand"],
            "model":  row["title_x"],   # ou 'title_y'
            "review": row["body"]
        })
    return {"results": results}

# démarre Uvicorn
uvicorn.run(app, host="0.0.0.0", port=8000)
```
